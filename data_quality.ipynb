```python
from pyspark.sql import functions as F


dim_cliente = spark.read.format("delta").load(
    "/Volumes/workspace/default/sources/delta/dim_cliente"
)

print("✔ DIM_CLIENTE")
dim_cliente.printSchema()

dim_cliente.select(
    F.count("*").alias("total_registros"),
    F.countDistinct("customer_id").alias("clientes_unicos"),
    F.sum(F.col("customer_id").isNull().cast("int")).alias("ids_nulos")
).show()
```

```python
dim_produto = spark.read.format("delta").load(
    "/Volumes/workspace/default/sources/delta/dim_produto"
)

print("✔ DIM_PRODUTO")

dim_produto.select(
    F.count("*").alias("total_registros"),
    F.countDistinct("product_id").alias("produtos_unicos"),
    F.sum(F.col("list_price").isNull().cast("int")).alias("preco_nulo")
).show()
```
